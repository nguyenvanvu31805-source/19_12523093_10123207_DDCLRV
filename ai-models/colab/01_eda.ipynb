```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="quality")

plt.title("Phân bố chất lượng rượu vang")
plt.xlabel("Quality")
plt.ylabel("Số lượng mẫu")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="quality")

plt.title("Phân bố chất lượng rượu vang")
plt.xlabel("Quality")
plt.ylabel("Số lượng mẫu")

plt.tight_layout()

plt.savefig("/content/01_quality_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="alcohol", kde=True)

plt.title("Phân bố nồng độ Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Số lượng mẫu")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="quality", y="alcohol")

plt.title("Mối quan hệ giữa Alcohol và Quality")
plt.xlabel("Quality")
plt.ylabel("Alcohol")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Ma trận tương quan giữa các biến")
plt.tight_layout()
plt.show()
```
